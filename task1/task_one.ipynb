## Download image from the remote path

````python
import urllib3
urladdress = 'https://wallup.net/wp-content/uploads/2015/06/Romantic-boy.jpg'
request = urllib3.request(method='GET', url=urladdress, headers={'User-Agent': 'Mozilla/5.0'})
print("Request status: "+ str(request.status))
````

## Get the image and print it
using ```display``` to print it within notebook

````python
from PIL import Image
from io import BytesIO
image = Image.open(BytesIO(request.data))
display(image)
````

### Image matrix

````python
import numpy as np
imageArray = np.asarray(image)
print(imageArray)
````

## Now convert it using Pillow library 


````python
image_h = image.size[1] /2
image_w = image.size[0] /2
newsize = (int(image_w), int(image_h))
newimage = image.resize(newsize).convert('L').rotate(90)
display(newimage)
````

### image array after image processing

````python
postimagearray = np.asarray(newimage)
print(postimagearray)
````
